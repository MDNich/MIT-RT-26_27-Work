```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

file = "runs_tab_angle_x/aero_results.csv"
df = pd.read_csv(file)
```

"X" Configuration


```python


colors = plt.cm.viridis(
    [i / (len(df["Beta"].unique()) - 1)
     for i in range(len(df["Beta"].unique()))]
)

line_styles = ["-", "--", ":"]

betas = sorted(df["Beta"].unique())
delta_es = sorted(df["delta_e"].unique())

for i, beta in enumerate(betas):

    for j, delta_e in enumerate(delta_es):

        data = df[
            (df["Beta"] == beta) &
            (df["delta_e"] == delta_e)
        ].sort_values("Alpha")

        plt.plot(
            data["Alpha"],
            data["CXtot"],
            color=colors[i],
            linestyle=line_styles[j],
            marker="o",
            label=f"β = {beta}°, δe = {delta_e}°"
        )

plt.xlabel("Alpha (deg)")
plt.ylabel("CXtot")
plt.title("CX vs Alpha for Beta and Elevator Deflection Sweeps")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
betas = sorted(df["Beta"].unique())
delta_es = sorted(df["delta_e"].unique())

colors = plt.cm.viridis(
    np.linspace(0, 1, len(betas))
)

line_styles = ["-", "--", ":"]

for i, beta in enumerate(betas):

    for j, delta_e in enumerate(delta_es):

        data = df[
            (df["Beta"] == beta) &
            (df["delta_e"] == delta_e)
        ].sort_values("Alpha")

        plt.plot(
            data["Alpha"],
            data["CDtot"],
            color=colors[i],
            linestyle=line_styles[j % len(line_styles)],
            marker="o",
            label=f"Beta = {beta}°, δe = {delta_e}°"
        )

plt.xlabel("Alpha (deg)")
plt.ylabel("CDtot")
plt.title("CD vs Alpha for Beta and Elevator Deflection Sweeps")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(True)
plt.tight_layout()
plt.show()
```
